```python
import glob
import json
from pathlib import Path

def find_missing_event_ids():
    # 1. Build a set of completed keys from the CSV file names
    csv_files = glob.glob("../Data/raw/event_results/*.csv")
    
    # Path(f).stem extracts just the base name without the .csv extension
    completed_keys = set(Path(f).stem for f in csv_files)
    
    # 2. Iterate through the JSON files to find missing keys
    json_files = glob.glob("../Data/raw/events/*.json")
    missing_event_ids = []
    
    for j_file in json_files:
        with open(j_file, 'r', encoding='utf-8') as f:
            try:
                data = json.load(f)
                event_key = data.get("Key")
                event_id = data.get("EventId")
                
                # If the key is valid but not found in our completed set, save the EventId
                if event_key and event_id is not None:
                    if event_key not in completed_keys:
                        # Store as string to easily join with commas later
                        missing_event_ids.append(str(event_id))
                        
            except json.JSONDecodeError:
                # Skip any files that are not valid JSON
                pass

    # 3. Output strictly the comma-separated list of integers
    # No headers, no tags, just the data
    return ",".join(missing_event_ids)

# Execute and print exactly what is requested
if __name__ == "__main__":
    result = find_missing_event_ids()
    print(result)
```
